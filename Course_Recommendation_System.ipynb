```python
import ast
```

```python
def remove_spaces(obj):
    l=[]
    if pd.isna(obj) or obj is None:
        return []
    for i in ast.literal_eval(obj):
        l.append(i.replace(' ',''))
    return l
```

```python
course_ratings['skills']=course_ratings['skills'].apply(remove_spaces)
```

```python
course_ratings['description']=course_ratings['description'].apply(lambda x:x.split())
```

```python
course_ratings['slug_x']=course_ratings['slug_x'].str.replace('-',' ').apply(lambda x:x.split())
```

```python
course_ratings['tags']=course_ratings['slug_x']+course_ratings['description']+course_ratings['skills']
```

```python
new_df=course_ratings[['url','final_title','tags','level','final_rating']]
new_df
```

```python
new_df['tags']=new_df['tags'].apply(lambda x: ' '.join(x))
```

```python
new_df['tags']=new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stemming(obj):
    y=[]
    for i in obj.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stemming)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
cv = TfidfVectorizer(max_features=5000, stop_words='english',ngram_range=(1,2))
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
def get_user_input():
    print("Welcome to the course recommeder!")
    user_skill=input("Which skill do you want to learn? (e.g., Python, Machine Learning, Leadership): ").lower()
    user_difficulty = input("2. What level of difficulty are you looking for? (Beginner, Intermediate, Mixed): ").lower()
    user_description = input("3. Describe the kind of course you want (e.g., 'practical applications', 'theoretical concepts', 'project-based'): ").lower()
    return user_skill,user_difficulty,user_description
```

```python
def query_vectorizer(user_skill,user_description,ps,cv):
    user_query=f"{user_skill} {user_skill} {user_skill} {user_description}"
    stemmed_query=stemming(user_query)
    user_vector=cv.transform([stemmed_query]).toarray()
    return user_vector
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend_courses(user_vector,new_df,user_difficulty,course_vectors=vectors,num_recs=5,rating_threshold=3.5):
    similarities = cosine_similarity(user_vector,course_vectors).flatten()
    sorted_indices=similarities.argsort()[::-1]
    recommendations=[]
    rec_count=0
    min_rating=new_df['final_rating'].min()
    max_rating = new_df['final_rating'].max()
    rating_boosts = ((new_df['final_rating'] - min_rating) / (max_rating - min_rating)).fillna(0) + 0.5
    print(f"\nSearching for {user_difficulty.capitalize()} courses...")
    print(f"Minimum rating considered: {rating_threshold}")
    for idx in sorted_indices:
        course_title=new_df.iloc[idx]['final_title']
        course_difficulty=new_df.iloc[idx]['level'].lower()
        course_rating = new_df.iloc[idx]['final_rating']
        raw_similarity = similarities[idx]
        if user_difficulty != 'mixed' and course_difficulty != user_difficulty:
            continue
        if course_rating < rating_threshold:
            continue
        final_score = (raw_similarity * 0.9) + (rating_boosts.iloc[idx] * 0.1)
        recommendations.append({
            'title': course_title,
            'url': new_df.iloc[idx]['url'],
            'difficulty': course_difficulty.capitalize(),
            'average_rating': course_rating,
            'similarity_score': raw_similarity,
            'final_ranking_score': final_score
        })
        rec_count += 1
        if rec_count >= num_recs:
            break
    recommendations = sorted(recommendations, key=lambda x: x['final_ranking_score'], reverse=True)
    if not recommendations:
        print("No courses found matching your criteria. Try different inputs.")
        return pd.DataFrame()

    return pd.DataFrame(recommendations)
```

```python
def recommendation_system():
    user_skill,user_difficulty,user_description=get_user_input()
    user_vector=query_vectorizer(user_skill,user_description,ps,cv)
    recommendations_df = recommend_courses(user_vector,new_df,user_difficulty)
    if not recommendations_df.empty:
        print("\n--- Your Recommended Courses ---")
        print(recommendations_df[['title', 'difficulty', 'average_rating', 'url', 'final_ranking_score']].to_string(index=False))
    else:
        print("Sorry, we couldn't find any recommendations based on your input.")
if __name__ == "__main__":
    recommendation_system()
```
